3. **Explain how the Scikit-Learn ColumnTransformer and Pipeline streamline the data preparation process and prevent data leakage across preprocessing and training steps.**

The column transformer and pipeline streamline the data preperation because they have the steps for standardization, encoding, and imputation in one neat object to be ran sequentially. It also allows for the seperate and parallel data preperation for numerical and categorical data. It prevents data leakage because the pipeline only gets the training data when pipeline.fit(X_train, y_train) is ran and has no access to the test set. If you were to impute and scale before the data is split then the mean or median will be generated partially based on test data.
