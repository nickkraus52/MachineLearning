# Dictionaries

```python
grade_dictionary = {"Joe" : 3.2, "Doug" : 2.8, "Mat" : 4.0}
for key, value in grade_dictionary.items():
    print(key, value)

print(grade_dictionary["Joe"]) # Prints 3.2
```

# Functions

```python
def absolute(num):
    if num < 0:
        return num * -1
    else:
        return num

num = absolute(-5)
print(num) # Prints 5
```
