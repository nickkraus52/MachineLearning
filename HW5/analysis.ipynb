2. **Contrast how hard voting aggregates predictions versus soft voting. Why do well-calibrated probability estimates matter when implementing soft voting across heterogeneous models?**

Hard voting takes the final classification of each model and gives each a vote. The majority vote is the classification of the ensamble. Soft voting takes a probability from each model and the overall probability is used to make a classification. In heterogeneous models each estimator is calibrated differently, and in a soft voting ensamble each model gets a vote so it is important each is well calibrated.

Because bagging is sampled with replacement about a third of the data will be ommited from the bagging, so the OOB score is created based on the data that is ommitted from the bagging. 4. **How does introducing random feature subsetting in Random Forests differ from standard Bagging, and how does this feature diversity impact model variance and generalization?**

Standard bagging trains on a random sample on a training set with replacement and allows consideration of all features on each decision tree. Feature subsetting on Random Forests allow decision trees to not use specific features to measure the impact of specific features and avoid having each decision tree always begin with splitting of a heavily weighted feature. The variety in decision tree feature differentiation helps generalization and decreases model variance.
